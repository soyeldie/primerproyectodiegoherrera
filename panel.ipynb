```python
import json

with open("data.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]
```

```python
# 1. Producto con mas ingresos
```

```python
def producto_mas_ingresos(ventas):
    ingresos = {}

    for venta in ventas:
        producto = venta["producto"]
        importe = venta["cantidad"] * venta["precio_unitario"]

        if producto not in ingresos:
            ingresos[producto] = 0

        ingresos[producto] += importe

    producto_ganador = max(ingresos, key=ingresos.get)

    return producto_ganador, ingresos[producto_ganador]


print("Producto con más ingresos:", producto_mas_ingresos(ventas))
```

```python
# 2. Ventas de Marta.

# Creo una lista vacia para almacenar las ventas de Marta y luego poder mostrarlo en una tabla porque solo el 
# resultado era una linea muy fea.
```

```python
def ventas_por_vendedor(ventas, vendedor):
    resultado = []

    for venta in ventas:
        if venta["vendedor"] == vendedor:
            resultado.append(venta)

    return resultado
```

```python
ventas_Marta = ventas_por_vendedor(ventas, "Marta")

print("VENTAS DE ANA")
print("-" * 75)
print(f"{'ID':<5} {'PRODUCTO':<20} {'CATEGORIA':<18} {'CANTIDAD':<10} {'PRECIO':<10}")
print("-" * 75)

for venta in ventas_Marta:
    print(f"{venta['id']:<5} {venta['producto']:<20} {venta['categoria']:<18} {venta['cantidad']:<10} {venta['precio_unitario']:<10}")

print("-" * 75)
```

```python
# 3. Ticket medio
```

```python
total = 0

for venta in ventas:
    total += venta["cantidad"] * venta["precio_unitario"]

ticket_medio = total / len(ventas)

print("Ticket medio:", round(ticket_medio, 2), "€")
```
